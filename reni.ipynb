```python
a = 108
eps_pox = 2^(-64)
prob = 3/8
prob1 = 1/2
prob2 = 1/4
```

```python
def renyi(a = a, eps = eps_pox, prob = 3/8, prob1 = 1/2, prob2 = 1/4):
    v = ( (1 - eps)*(prob)**-a + (eps*(2/5) )*(prob1)**-a + (eps*(3/5) )*(prob2)**-a)^(1/a) 
    vclose = v -  prob^-1
    return log(vclose,2.)
```
